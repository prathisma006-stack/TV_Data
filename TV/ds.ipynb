```python
df.loc[
    df["Subscription_Fee"] < 0,
    "Subscription_Fee"
] = np.nan
```

```python
df["Subscription_Fee"] = df["Subscription_Fee"].fillna(
    df["Subscription_Fee"].median()
)
```

```python
df["Avg_Viewers"] = pd.to_numeric(
    df["Avg_Viewers"],
    errors="coerce"
)
```

```python
df["Avg_Viewers"].dtype
```

```python
Q1 = df["Avg_Viewers"].quantile(0.25)
Q3 = df["Avg_Viewers"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Limit:", lower)
print("Upper Limit:", upper)
```

```python
outliers = df[
    (df["Avg_Viewers"] < lower) |
    (df["Avg_Viewers"] > upper)
]

outliers
```

```python
import matplotlib.pyplot as plt
```

```python
df["Language"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Number of TV Channels by Language")
plt.xlabel("Language")
plt.ylabel("Number of Channels")

plt.show()
```

```python
df["HD_Available"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7,7)
)

plt.title("HD Availability")
plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["Avg_Viewers"],
    bins=30
)

plt.title("Distribution of Average Viewers")
plt.xlabel("Average Viewers")
plt.ylabel("Number of Channels")

plt.show()
```
